```python
import zipfile
import os
import pandas as pd

zip_path = "/content/air+quality.zip"
extract_path = "/content/air_quality_data"

with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall(extract_path)

csv_path = os.path.join(extract_path, "AirQualityUCI.csv")

df = pd.read_csv(csv_path, sep=";", decimal=",")
df = df.dropna(axis=1, how="all")

print("DATASET LOADED SUCCESSFULLY")
print("=" * 50)
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nFirst 5 rows:")
display(df.head())
print("\nDataset information:")
df.info()
print("\nStatistical summary:")
display(df.describe(include="all").T)
```

```python
import numpy as np
import pandas as pd

# CLEAN DATA
print("BEFORE CLEANING")
print("Shape:", df.shape)
print("Duplicates:", df.duplicated().sum())

df = df.drop_duplicates().copy()
df = df.dropna(how="all")

numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols] = df[numeric_cols].replace(-200, np.nan)

# Remove feature with excessive missing values
df = df.drop(columns=["NMHC(GT)"])

# Create datetime feature
df["DateTime"] = pd.to_datetime(
    df["Date"] + " " + df["Time"].str.replace(".", ":", regex=False),
    dayfirst=True,
    errors="coerce"
)

df = df.drop(columns=["Date", "Time"])
df = df.dropna(subset=["DateTime", "CO(GT)"])
df = df.sort_values("DateTime").reset_index(drop=True)

# Impute missing sensor features
feature_cols = df.select_dtypes(include=np.number).columns.drop("CO(GT)")
df[feature_cols] = df[feature_cols].interpolate(limit_direction="both")
df[feature_cols] = df[feature_cols].fillna(df[feature_cols].median())

# Time-based features
df["Hour"] = df["DateTime"].dt.hour
df["Month"] = df["DateTime"].dt.month
df["DayOfWeek"] = df["DateTime"].dt.dayofweek

df = df.drop(columns=["DateTime"])
df = df.dropna()

print("\nAFTER CLEANING")
print("Shape:", df.shape)
print("Duplicates:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())
print("\nFinal columns:", df.columns.tolist())
display(df.head())
```

```python
# FEATURE SCALING AND TRAIN-TEST SPLIT

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import time

X = df.drop(columns=["CO(GT)"])
y = df["CO(GT)"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("TRAINING SAMPLES:", len(X_train))
print("TESTING SAMPLES:", len(X_test))
print("FEATURES:", X.shape[1])
print("SCALING COMPLETE")

# TRAIN AND COMPARE MODELS

models = {
    "Linear Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LinearRegression())
    ]),
    "Random Forest": RandomForestRegressor(
        n_estimators=100, max_depth=15,
        random_state=42, n_jobs=-1
    ),
    "Gradient Boosting": HistGradientBoostingRegressor(
        max_iter=150, random_state=42
    )
}

results = []
trained_models = {}

for name, model in models.items():
    start = time.time()
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    elapsed = time.time() - start

    results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_test, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_test, predictions)),
        "R2 Score": r2_score(y_test, predictions),
        "Training Time (s)": round(elapsed, 2)
    })
    trained_models[name] = model

results_df = pd.DataFrame(results).sort_values("RMSE")

print("MODEL COMPARISON")
display(results_df)

best_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_name]

print("Selected Model:", best_name)
print("Best Test RMSE:", round(results_df.iloc[0]["RMSE"], 4))
```

```python
import os
import json
import joblib
import zipfile
import numpy as np

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

PROJECT_DIR = "/content/air_quality_project"
os.makedirs(PROJECT_DIR, exist_ok=True)

# 1. Save model and scaler
joblib.dump(model, f"{PROJECT_DIR}/air_quality_model.joblib")
joblib.dump(scaler, f"{PROJECT_DIR}/scaler.joblib")

# 2. Save feature names
feature_names = list(X.columns) if hasattr(X, "columns") else [
    "PT08.S1(CO)", "C6H6(GT)", "PT08.S2(NMHC)", "NOx(GT)",
    "PT08.S3(NOx)", "NO2(GT)", "PT08.S4(NO2)", "PT08.S5(O3)",
    "T", "RH", "AH", "Hour", "Month", "DayOfWeek"
]

with open(f"{PROJECT_DIR}/feature_names.json", "w") as f:
    json.dump(feature_names, f, indent=4)

# 3. Create CLI prototype
cli_code = '''
import joblib
import json
import numpy as np

path = "/content/air_quality_project"

model = joblib.load(f"{path}/air_quality_model.joblib")
scaler = joblib.load(f"{path}/scaler.joblib")

with open(f"{path}/feature_names.json") as f:
    features = json.load(f)

print("\\\\nAIR QUALITY PREDICTION")
print("-" * 35)

values = []
for feature in features:
    value = float(input(f"Enter {feature}: "))
    values.append(value)

data = np.array(values).reshape(1, -1)
data_scaled = scaler.transform(data)
prediction = model.predict(data_scaled)[0]

print(f"\\\\nPredicted CO(GT): {prediction:.4f}")
'''

with open(f"{PROJECT_DIR}/cli.py", "w") as f:
    f.write(cli_code)

# 4. Create project summary
summary = """
AIR QUALITY PREDICTION PROJECT

Dataset:
UCI Machine Learning Repository - Air Quality dataset.

Objective:
Predict carbon monoxide concentration (CO(GT)) using air-quality
sensor measurements, weather variables, and time features.

Preprocessing:
- Removed duplicate records.
- Handled missing values.
- Replaced invalid -200 measurements.
- Extracted hour, month, and day-of-week features.
- Scaled input features.

Models compared:
- Linear Regression
- Random Forest
- Gradient Boosting

Best model:
Linear Regression

Test results:
MAE: 0.3196
RMSE: 0.4837
R2 Score: 0.8706

Prototype:
cli.py accepts feature values and predicts CO(GT).

Note:
Predictions depend on the quality and units of the input measurements.
"""

with open(f"{PROJECT_DIR}/project_summary.txt", "w") as f:
    f.write(summary.strip())

# 5. Package project artifacts
zip_path = "/content/air_quality_project.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for filename in os.listdir(PROJECT_DIR):
        z.write(
            os.path.join(PROJECT_DIR, filename),
            arcname=filename
        )

print("PROJECT FILES CREATED")
print(os.listdir(PROJECT_DIR))
print(f"\nZIP FILE: {zip_path}")

# 6. Download project ZIP
from google.colab import files
files.download(zip_path)
```
